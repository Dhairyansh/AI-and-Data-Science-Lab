# question 1

i have entered a integer value through which it has clculated mean and variance using numpy library 

```python
import numpy as np
n = int(input("random number: "))
if(n<=0):
    print("enter a positive natural number")
arr = np.random.rand(n)
mean = np.mean(arr)
variance = np.var(arr)
print("random array:" , arr)
print("mean:", mean)
print("variance:", variance)
```

# question 2

i have given the value of n and x then i have used the while/for loop to calcuate the sum 

```python
import numpy as np
n = int(input("enter the value of n: "))
x = int(input("enter the value of x: "))
sum1=0
sum2=0
j = 1
for i in range(1,n+1):
    sum1 += (((-1)**(i))*(x**(i/2)))/((i)*(i+1))
print(sum1) 
while j<=n:
    sum2 += (((-1)**(j))*(x**(j/2)))/((j)*(j+1))
    j += 1
print(sum2)    
```

```python
import matplotlib.pyplot as plt
import csv
import numpy as np
def f1(x):
    return (x**2 + 2*x + 3)/(x + 20)
def f2(x):
    return x**3 - 6*x**2 + 4*x - 1
def f3(x):
    return np.exp(x) - 1 + x**2 -x**3
def f4():
    return 2*(np.sin(x)) + x/3 -4

functions = {
    "1": (f1, "f1(x) = (x^2 + 2x + 3)/(x + 20)"),
    "2": (f2, "f2(x) = x^3 - 6x^2 + 4x - 1"),
    "3": (f3, "f3(x) = e^x - 1 + x^2 - x^3"),
    "4": (f4, "f4(x) = 2sin(x) + x/3 - 4")
}

choice = input("Choose function (1,2,3,4): ")
xmin = float(input("Enter xmin: "))
xmax = float(input("Enter xmax: "))
n = int(input("Enter number of points n: "))

if choice in functions:
    func, label = functions[choice]
    x_vals = np.linspace(xmin, xmax, n)
    y_vals = func(x_vals)
    plt.plot(x_vals, y_vals, label=label, color="blue")
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.title(f"Plot of {label}")
    plt.legend()
    plt.grid(True)
    plt.show()
    
    with open("function_values.csv", "w", newline="") as file:
           writer = csv.writer(file)
           writer.writerow(["x", "f(x)"])  
           for xi, yi in zip(x_vals, y_vals):
             writer.writerow([xi, yi])

    print("Data saved to function_values.csv")

```
