# in this question I have made a function which calculate the value of integral through trapezodial's rule which calls and function to complete the calculation. And a simpson fuction with two FOR loops to calculate the integral(only if n is even) .

```python
import math

def trap(func, a, b, n):
    h = (b - a) / n
    result = 0
    i = 1
    while i < n:
        result += func(a + i * h)
        i += 1
    return result  

def integral(func, a, b, n,):
    h = (b - a) / n
    result = (h/2) * (func(a) + 2 * trap(func, a, b, n) + func(b))
    return result

n = int(input("Enter the value of n: "))

def f_a(x):
    return math.sin(x)

def f_b(x):
    return math.exp(-x)

def f_c(x):
    return 1 / (1 + x**2)

def f_d(x):
    return 1 / x

def f_e(t):
    return math.sqrt(1 - 0.25 * (math.sin(t)**2))

def f_f(x):
    return 1 / (math.sin(x)**2 + 0.25*math.cos(x)**2)

res_a = integral(f_a, 0, math.pi/2, n)
res_b = integral(f_b, 0, 2, n)
res_c = integral(f_c, 0, 1, n)
res_d = integral(f_d, 1, 2, n)
res_e = integral(f_e, 0, math.pi/2, n)
res_f = integral(f_f, 0, math.pi/2, n)

print(" trapezoidal’s Rule")
print("(a) =", res_a, " exact = 1")
print("(b) =", res_b, " exact ≈ 0.86466")
print("(c) =", res_c, " exact =", math.pi/4)
print("(d) =", res_d, " exact =", math.log(2))
print("(e) =", res_e, " exact ≈ 1.4675")
print("(f) =", res_f, " exact =", math.pi)

if(n%2== 0):
    def integral_simpson(func, a, b, n):    
       h = (b - a) / n
       result = func(a) + func(b)
    
       for i in range(1, n, 2):
          result += 4 * func(a + i*h)

       for i in range(2, n-1, 2):
          result += 2 * func(a + i*h)

       return (h/3) * result

    res_a_simp = integral_simpson(f_a, 0, math.pi/2, n)
    res_b_simp = integral_simpson(f_b, 0, 2, n)
    res_c_simp = integral_simpson(f_c, 0, 1, n)
    res_d_simp = integral_simpson(f_d, 1, 2, n)
    res_e_simp = integral_simpson(f_e, 0, math.pi/2, n)
    res_f_simp = integral_simpson(f_f, 0, math.pi/2, n)

    print(" Simpson’s Rule")
    print("(a) =", res_a_simp, " exact = 1")
    print("(b) =", res_b_simp, " exact ≈ 0.86466")
    print("(c) =", res_c_simp, " exact =", math.pi/4)
    print("(d) =", res_d_simp, " exact =", math.log(2))
    print("(e) =", res_e_simp, " exact ≈ 1.4675")
    print("(f) =", res_f_simp, " exact =", math.pi)

else:
    print("please enter a even integer")
```

# in this question I have plotted this graph using iris.csv 

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("iris.csv")

print("Available columns for plotting:")
print("1. sepal-length")
print("2. sepal-width")
print("3. petal-length")
print("4. petal-width")

x_col = input("Enter the column name you want on the x-axis: ")
y_col = input("Enter the column name you want on the y-axis: ")


species_list = data["species"].unique()   
colors = ["blue", "orange", "green"]     

for i in range(len(species_list)):
    species_name = species_list[i]
    color = colors[i]
    
    subset = data[data["species"] == species_name]
    
    plt.scatter(subset[x_col], subset[y_col], color=color, label=species_name)

plt.xlabel(x_col + " (cm)")
plt.ylabel(y_col + " (cm)")
plt.title("Scatter Plot of " + x_col + " vs " + y_col)
plt.legend()
plt.grid(True)

plt.show()
```
