# question 1: in this question I have plotted a histogram by giving inputs of std-var, mean, random number N.

```python
import numpy as np
import matplotlib.pyplot as plt
N = int(input("enter the number of random number:"))
mumu = int(input("enter the mean:"))
chigma = int(input("enter the standard deviation:"))

data = np.random.normal(mumu, chigma, N)

plt.hist(data, bins=100, edgecolor='black', alpha=0.7)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title(f"Histogram of Normally Distributed Data (mean={mumu}, std={chigma})")

plt.show()
```

# question 2: in this question I have removed the rows with question mark in the fourth row and then
# calculated the mean, median, standard deviation, skewness and kurtosis the column three and four and then 
# I have plotted the histogram of these two column's data. 

```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis

def mean(data):
    return sum(data) / len(data)

def median(data):
    data_sorted = sorted(data)
    n = len(data_sorted)
    mid = n // 2
    if n % 2 == 0:  
        return (data_sorted[mid - 1] + data_sorted[mid]) / 2
    else:           
        return data_sorted[mid]

def standard_deviation(data):
    m = mean(data)
    variance = sum((x - m) ** 2 for x in data) / len(data)
    return variance ** 0.5

def skewness(data):
    m = mean(data)
    sd = standard_deviation(data)
    n = len(data)
    return sum((x - m) ** 3 for x in data) / (n * (sd ** 3))

def kurtosis(data):
    m = mean(data)
    sd = standard_deviation(data)
    n = len(data)
    return (sum((x - m) ** 4 for x in data) / (n * (sd ** 4))) - 3

df = pd.read_csv("auto.csv")

df_cleaned = df[df.iloc[:, 3] != '?']

df_cleaned.to_csv("auto_cleaned.csv", index=False)

df = pd.read_csv("auto_cleaned.csv")

column_data = df.iloc[:, 2]

column_data = pd.to_numeric(column_data)

data_list = column_data.tolist()

print("by formula")
print("Mean:", mean(data_list))
print("Median:", median(data_list))
print("Standard Deviation:", standard_deviation(data_list))
print("Skewness:", skewness(data_list))
print("Kurtosis:", kurtosis(data_list))

print("by scipy and numpy")
print("Mean:", np.mean(data_list))
print("Median:", np.median(data_list))
print("Standard Deviation:", np.std(data_list))
print("Skewness:", skew(data_list))
print("Kurtosis:", kurtosis(data_list))

df = pd.read_csv("auto_cleaned.csv")
col3 = df.iloc[:, 2]
col4 = df.iloc[:, 3]


plt.hist(col3, bins=30, edgecolor='black', alpha=0.7)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title(f"Histogram of the displacement (mean={mean(col3)}, std={standard_deviation(col3)})")

plt.show()

plt.hist(col4, bins=30, edgecolor='black', alpha=0.7, color = 'red')

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title(f"Histogram of the horsepower (mean={mean(col4)}, std={standard_deviation(col4)})")

plt.show()
```

# question 3: in this I have followed the steps of the question and then plotted the histogram respectively

```python
import numpy as np
import matplotlib.pyplot as plt

N = 100000
a, b = 0, 100
pop = np.random.uniform(a, b, N)
mu = (a + b) / 2
sigma = ((b - a) ** 2 / 12) ** 0.5

m = 30
k = 10000
means = []
for i in range(k):
    sample = np.random.choice(pop, m)
    means.append(np.mean(sample))

plt.hist(pop, bins=50, edgecolor='black')
plt.title("Population Histogram")
plt.show()

plt.hist(means, bins=50, edgecolor='black', color = 'red')
plt.title("Sample Means Histogram")
plt.show()

print("Population mean:", mu)
print("Population std:", sigma)
print("Sample means mean:", np.mean(means))
print("Sample means std:", np.std(means, ddof=1))
print("Theoretical mean of sample means:", mu)
print("Theoretical std of sample means:", sigma / (m ** 0.5))
```
