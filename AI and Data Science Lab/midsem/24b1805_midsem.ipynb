```python
import numpy as np
import matplotlib.pyplot as plt

mu1, sigma1 = 2.0, 1.0   
mu2, sigma2 = -1.0, 0.5    

x1 = np.random.normal(mu1, sigma1, 100)  
x2 = np.random.normal(mu2, sigma2, 100) 

data = np.column_stack((x1, x2))

print("Generated 100x2 array:\n", data)

plt.scatter(x1, x2, c='blue', alpha=0.6, edgecolors='k')
plt.title("Scatter plot of (x1, x2)")
plt.xlabel("x1 ~ N(2.0, 1.0^2)")
plt.ylabel("x2 ~ N(-1.0, 0.5^2)")
plt.grid(True)
plt.show()
```

# question 2

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def experiment(n, num_samples=10000, lam=1.0):
    
    samples = np.random.exponential(scale=1/lam, size=(num_samples, n))
    
    sample_means = samples.mean(axis=1)
    
    plt.hist(sample_means, bins=50, density=True, alpha=0.6, color='skyblue', edgecolor='black')
    
    mu = 1/lam                
    sigma = 1.0               
    sigma_x = sigma / np.sqrt(n)  
    
    x = np.linspace(min(sample_means), max(sample_means), 200)
    plt.plot(x, norm.pdf(x, mu, sigma_x), 'r-', lw=2, label='Normal Approximation')
    
    plt.title(f"Distribution of Sample Means (n={n})")
    plt.xlabel("Sample mean")
    plt.ylabel("Density")
    plt.legend()
    plt.grid(True)
    plt.show()
    
    return sample_means


for n in [5, 30, 100]:
    experiment(n)
```

# n = 5 (small sample size):
   # 1. The histogram of sample means is still right-skewed (since the exponential distribution itself is skewed).
   # 2. The spread is wide, meaning the sample means vary a lot.
   # 3. The normal curve overlay doesn’t fit well.
# n = 30 (moderate sample size):
   # 1. The histogram looks much closer to a bell shape.
   # 2. Skewness is reduced compared to n=5.
   # 3. The sample mean distribution begins to align well with the normal approximation.
   # 4.This is consistent with the Central Limit Theorem (CLT), which says that the distribution of sample means tends toward normality as n increases.
# n = 100 (large sample size):
   # 1. The histogram is very close to a normal distribution.
   # 2. The spread is narrower (since standard error = sigma/(sqrt n)).
   # 3. The normal curve matches the histogram almost perfectly.
   # 4. This shows that with larger samples, the sample mean is a very reliable estimator of the population mean.
# Discussion:
   # 1. Shape: As n increases, the skewness decreases and the distribution becomes more symmetric and normal-like.
   # 2. Spread: The variability of sample means shrinks as n grows, because standard error = sigma/(sqrt n) gets smaller
   # 3. Center: The mean of the sample means always centers around the population mean (mu=1.0), regardless of n.

# question 3

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import skew, kurtosis 

df = pd.read_csv("student_scores.csv")

col_choice = input("Enter column to analyze (TestScoreA or TestScoreB): ").strip()

if col_choice not in df.columns:
    print("Invalid choice. Please enter either 'TestScoreA' or 'TestScoreB'.")
    exit()

data = df[col_choice].dropna().tolist()  
n = len(data)

mean = 0
for x in data:
    mean += x
mean /= n

var = 0
for x in data:
    var += (x - mean) ** 2
var /= n
std_dev = var ** 0.5

skewness_num = 0
for x in data:
    skewness_num += (x - mean) ** 3
skewness = (skewness_num / n) / (std_dev ** 3)

kurtosis_num = 0
for x in data:
    kurtosis_num += (x - mean) ** 4
kurt = (kurtosis_num / n) / (std_dev ** 4)

print(f"\nAnalysis of {col_choice}:")
print(f"Mean (µ)             = {mean:.4f}")
print(f"Std Dev (σ)          = {std_dev:.4f}")
print(f"Skewness             = {skewness:.4f}")
print(f"Kurtosis(full)       = {kurt:.4f}")

print("\nVerification with numpy/scipy:")
print("Mean:", np.mean(data))
print("Std Dev:", np.std(data))
print("Skewness:", skew(data))
print("Kurtosis(full):", kurtosis(data, fisher=False))  

plt.hist(data, bins=30, edgecolor="black", alpha=0.7)
plt.title(f"Histogram of {col_choice}")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
from scipy.stats import norm, ttest_1samp
import random

df = pd.read_csv("student_scores.csv")

col_choice = input("Enter column to analyze (TestScoreA or TestScoreB): ").strip()

if col_choice not in df.columns:
    print("Invalid column name. Please enter TestScoreA or TestScoreB.")
    exit()

data = df[col_choice].dropna().tolist()
N = len(data)  

mu_0 = np.mean(data)
sigma = np.std(data, ddof=0) 

test_choice = input("Enter test type (z-test or t-test): ").strip().lower()

n = int(input("Enter sample size n (<= 800): "))

sample = random.sample(data, n)

alpha = float(input("Enter significance level alpha (e.g., 0.05): "))

sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)

if test_choice == "z-test":
    z_stat = (sample_mean - mu_0) / (sigma / np.sqrt(n))
    p_value = 2 * (1 - norm.cdf(abs(z_stat)))
    print(f"\nZ statistic = {z_stat:.4f}, p-value = {p_value:.4f}")

elif test_choice == "t-test":
    t_stat, p_value = ttest_1samp(sample, mu_0)
    print(f"\nT statistic = {t_stat:.4f}, p-value = {p_value:.4f}")

else:
    print("Invalid test choice. Choose either 'z-test' or 't-test'.")
    exit()

if p_value < alpha:
    print("Conclusion: Reject H0 (two-tailed test).")
else:
    print("Conclusion: Fail to reject H0 (two-tailed test).")
```
